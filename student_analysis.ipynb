```python
import matplotlib.pyplot as plt
plt.bar(df["Name"],df["Maths"])
plt.title("Maths Marks of Students")
plt.xlabel("Students")
plt.ylabel("Maths Marks")
plt.show()
```

```python
df["Total"]=df["Maths"]+df["Chemistry"]+df["English"]
df
```

```python
df["Average"]=df["Total"]/3
df
```

```python
df[df["Average"]==df["Average"].max()]
```

```python
df.sort_values(by="Average",ascending=False)
```

```python
df.to_csv("student_analysis_results.csv",index=False)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8,5))
plt.bar(df["Name"],df["Average"])
plt.title("Average Marks of Students")
plt.xlabel("Student Name")
plt.ylabel("Average Marks")
plt.xticks(rotation=45)
plt.savefig("average_marks_chart.png")
plt.show()
```

```python
subject_average=df[["Maths","Chemistry","English"]].mean()
subject_average.plot(kind="bar")
plt.title("Average Marks of Subject")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.show()
```

```python
import matplotlib.pyplot as plt
subject_average=df[["Maths","Chemistry","English"]].mean()
plt.figure(figsize=(6,6))
plt.pie(subject_average,labels=subject_average.index,autopct="%1.1f%%",startangle=90)
plt.title("Average Marks Distribution by Subject")
plt.show()
```
